## data Cleaning

```python
df['TotalCharges'].dtype
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'],errors = 'coerce')
```

```python
df['TotalCharges'].fillna(df['TotalCharges'].median(),inplace = True)
```

```python
target_map = {'Yes': 1, 'No': 0}
df['Churn'] = df['Churn'].map(target_map)
```

```python
df_le = df.copy()
```

```python
categorical_cols = df.select_dtypes(include = ['object']).columns
```

```python
binary_cat_cols = [col for col in categorical_cols if df_le[col].nunique()==2]
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in binary_cat_cols:
    if df_le[col].nunique() == 2:
        df_le[col] = le.fit_transform(df_le[col])
```

```python
df_le.head()
```

## Target Encoding

```python
non_binary_cols = [col for col in categorical_cols if df_le[col].nunique() > 2]
```

```python
non_binary_cols
```

```python
train_df,test_df = train_test_split(df_le,test_size=0.2,random_state=101)
```

```python
test_df.head()
```

```python
train_df.head()
```

```python
def target_encoding(train_df, test_df, cols):
    for col in cols:
        mean_map = train_df.groupby(col)['Churn'].mean()
        train_df[col] = train_df[col].map(mean_map)
        test_df[col] = test_df[col].map(mean_map)
        
    return train_df, test_df
```

```python
train_df_encoded, test_df_encoded = target_encoding(train_df, test_df, non_binary_cols)
```
